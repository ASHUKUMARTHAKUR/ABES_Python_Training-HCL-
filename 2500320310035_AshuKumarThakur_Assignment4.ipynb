```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns
import plotly.express as px
from IPython.display import display, Image

sales = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Laptops":    [40, 42, 45, 38, 36, 35, 41, 48, 52, 85, 95, 60],
    "Phones":     [55, 54, 58, 60, 57, 59, 62, 64, 66, 90, 98, 72],
    "Headphones": [20, 22, 21, 25, 24, 26, 25, 27, 30, 45, 52, 40],
})

# --TASK 1--

plt.figure()
for product in ["Laptops", "Phones", "Headphones"]:

    plt.plot(sales["Month"], sales[product], label=product)
plt.legend()
plt.savefig("before.png")

plt.show()


```

```python
# --TASK 5--
import pandas as pd
import matplotlib.pyplot as plt


months = ["Jan", "Feb", "Mar", "Apr", "May", "June", "July", "Aug", "Sep", "Oct", "Nov", "Dec"]
phone_sales = [55, 54, 58, 60, 57, 59, 62, 64, 66, 90, 72, 80]

sales = pd.DataFrame({"Phones": phone_sales}, index=months)

max_month = sales["Phones"].idxmax()

colors = ["crimson" if idx == max_month else "lightgrey" for idx in sales.index]

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.bar(sales.index, sales["Phones"], color=colors)


ax.bar_label(bars, padding=3)
ax.set_title("Phone Sales by Month (Peak Highlighted in Crimson)", fontsize=14)
ax.set_xlabel("Month")
ax.set_ylabel("Sales Units")
ax.spines[["top", "right"]].set_visible(False)

plt.show()
```

```python
# --TASK 6--



long = sales.melt(
    id_vars="Month",
    var_name="Product",
    value_name="Sales"
)


fig = px.line(
    long,
    x="Month",
    y="Sales",
    color="Product",
    markers=True,
    color_discrete_map=colors,
    template="simple_white",
    title="Phones lead all year, and every product peaks in November"
)


fig.update_traces(
    line_width=3,
    hovertemplate="%{x}: Rs. %{y}k"
)


fig.update_layout(
    yaxis_title="Sales (Rs. thousands)"
)


fig.add_annotation(
    x="Nov",
    y=98,
    text="Festive season peak",
    showarrow=True,
    arrowhead=2
)


fig.add_vrect(
    x0="Oct",
    x1="Nov",
    fillcolor="orange",
    opacity=0.2,
    line_width=0
)


fig.show()
```
