```python
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
monthly_sales = [25000, 30000, 28000, 32000, 40000, 45000]

# Slicing months to match the length of monthly_sales (6 months)
plt.plot(months[:len(monthly_sales)], monthly_sales, marker="o", color="blue")

plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.title("Monthly Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt

regions = ["North", "South", "East", "West", "Central"]
sales = [50000, 65000, 40000, 70000, 55000]

plt.bar(regions, sales, color="red")
plt.xlabel("Regions")
plt.ylabel("Sales")
plt.title("Sales by Region")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

tips = px.data.tips()
tips.head()
import matplotlib.pyplot as plt
region=["North", "South", "East", "West"]
plt.bar(region, [100, 150, 200, 250], color="blue")
sales_data = [10000, 10500, 200000, 25000]
plt.plot(region, sales_data, marker="o", color="green")
plt.title("Sales Data by Region")
plt.xlabel("Region")
plt.ylabel("Sales ($)")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ["Electronics", "Clothing", "Grocery", "Furniture", "Sports" ]
sales = [45000, 30000, 20000, 15000, 10000]

plt.figure(figsize=(7, 7))

plt.pie( sales, labels = categories, autopct='%1.1f%%')
```

```python
import matplotlib.pyplot as plt

salaries=[25000, 30000, 28000, 32000, 40000, 45000,
          20000, 35000, 25000, 38000, 42000, 47000,
          18000, 28000, 36000, 41000, 39000, 50000 ]

plt.figure(figsize=(6, 5))

plt.hist(
        salaries,
        bins = 5,
        color ="skyblue",
        edgecolor = "black"
)

plt.title("Employee Salary Distribution")
plt.xlabel("Monthly Salary (Rs.)")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
it_salary = [40000, 45000, 50000, 55000, 60000]
hr_salary = [30000, 35000, 40000, 42000, 45000]
finance_salary = [35000, 40000, 45000, 50000, 55000]

salary_data = [it_salary, hr_salary, finance_salary]
plt.figure(figsize=(8, 5))

plt.boxplot(salary_data, tick_labels=["IT", "HR", "Finance"])

plt.title("Salary Distribution")
plt.xlabel("Category")
plt.ylabel("Salary")
plt.show()
```
