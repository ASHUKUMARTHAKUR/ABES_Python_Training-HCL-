```python
# --TASK 1

import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
from IPython.display import display, Image


sales = pd.DataFrame({
    "Month":["Jan", "Feb", "Mar", "Apr", "May", "June", "July", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Laptops":[40, 42, 45, 38, 36, 35, 41, 48, 52, 85, 95, 60],
    "Phones":[55, 54, 58, 60, 57, 59, 62, 64, 66, 90, 72, 80],
    "Headphones":[20, 22, 21, 25, 24, 26, 25, 27, 30, 45, 52, 40]
 })


plt.figure()
for product in ["Laptops", "Phones", "Headphones"]:

    plt.plot(sales["Month"], sales[product], label=product)
plt.legend()
plt.savefig("before.png")

plt.show()
```

```python
# --TASK 2

sales = pd.DataFrame({
    "Month":["Jan", "Feb", "Mar", "Apr", "May", "June", "July", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Laptops":[40, 42, 45, 38, 36, 35, 41, 48, 52, 85, 95, 60],
    "Phones":[55, 54, 58, 60, 57, 59, 62, 64, 66, 90, 72, 80],
    "Headphones":[20, 22, 21, 25, 24, 26, 25, 27, 30, 45, 52, 40]
 })

print(sales)

plt.style.use("seaborn-v0_8-whitegrid")
colors = {
    "Laptops":"#0072B2",
    "Phones":"#D55E00",
    "Headphones":"#009E73"
}

fig, ax = plt.subplots(figsize=(10, 6))

for product in colors:
  ax.plot(
    sales["Month"],
    sales[product],
    label=product,
    color=colors[product],
    linewidth=2.5,
    marker="o"
    )


plt.legend()
plt.show()
```

```python
#--Task-3--
sales = pd.DataFrame({
    "Month":["Jan", "Feb", "Mar", "Apr", "May", "June", "July", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Laptops":[40, 42, 45, 38, 36, 35, 41, 48, 52, 85, 95, 60],
    "Phones":[55, 54, 58, 60, 57, 59, 62, 64, 66, 90, 72, 80],
    "Headphones":[20, 22, 21, 25, 24, 26, 25, 27, 30, 45, 52, 40]
 })
colors={
    "Laptops":"red",
    "Phones":"blue",
    "Headphones":"green"
}
fig,ax=plt.subplots(figsize=(9,6))
for product in colors:
    ax.plot(sales["Month"],sales[product],label=product,color=colors[product],linewidth=2.5,marker="o")
    ax.text(11.2,
            sales[product].iloc[-1],
            product,
            color=colors[product],
            fontsize=11,
            va="center")

ax.set_title("phone lead all year,and every product peak in nov")
ax.set_xlabel("month")
ax.set_ylabel("sales")
ax.set_xlim(-0.5,13)
ax.spines[["top","right"]].set_visible(False)
plt.show()
```

```python
#--Task-4--


sales = pd.DataFrame({
    "Month":["Jan", "Feb", "Mar", "Apr", "May", "June", "July", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Laptops":[40, 42, 45, 38, 36, 35, 41, 48, 52, 85, 95, 60],
    "Phones":[55, 54, 58, 60, 57, 59, 62, 64, 66, 90, 72, 80],
    "Headphones":[20, 22, 21, 25, 24, 26, 25, 27, 30, 45, 52, 40]
 })
colors={
    "Laptops":"red",
    "Phones":"blue",
    "Headphones":"green"
}
fig,ax=plt.subplots(figsize=(11,6))
for product in colors:
    ax.plot(sales["Month"],sales[product],label=product,color=colors[product],linewidth=2.5,marker="o")
    ax.text(11.2,
            sales[product].iloc[-1],
            product,
            color=colors[product],
            fontsize=11,
            va="center")

ax.set_title("phone lead all year,and every product peak in nov",
             fontsize=14,
             )
ax.set_xlabel("month")
ax.set_ylabel("sales")
ax.axvspan(8.5,10.5,color="orange",alpha=0.2)
ax.annotate(
    "festival peak",
    xy=(10,98),
    xytext=(5.5,90),
    arrowprops={"facecolor":"orange"}
)
ax.text(
    0.2,
    62,
    "monthly "
)
ax.set_xlim(-0.5,13)
ax.spines[["top","right"]].set_visible(False)
fig.savefig("after.png",dpi=150)
plt.show()
```
